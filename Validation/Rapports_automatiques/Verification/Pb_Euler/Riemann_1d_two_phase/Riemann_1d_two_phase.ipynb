```python
def plot_baltik(list_meshes,list_tests):
    import pandas as pd
    from trustutils import plot
    variables = [ "ALPHA_SOLIDE", "ALPHA_GAZ", "RHO_SOLIDE", "RHO_GAZ", "V_SOLIDE", "V_GAZ", "P_SOLIDE","P_GAZ"]
    dis_m = {"Coloc":"o"}
    for test in list_tests:
        df = plot.read_csv(f"{run.BUILD_DIRECTORY}/Exacts/{test}.ex", sep=r'\s+', usecols=[0,1,2,3,4,5,6,7,8])
        a = plot.Graph(nX=3, nY=3 , title=test)
        nX, nY = 0, 0
        for i, p in enumerate(variables):
         nY = i % 3     
         nX = i // 3
         a.addPlot([nX, nY], f" ")
         for m in list_meshes:
            for dis, ma in dis_m.items():
                if dis == "Coloc":
                    a.addSegment(f"n{m}/{dis}/{test}_{p}.son", label=f"{dis} n={m}",lw = 2, marker=ma)
         a.add(df["x"], df[f"{p}"], label="Exact solution", lw=1, color='k', marker="--") 

plot_baltik(meshes,tests)
```
