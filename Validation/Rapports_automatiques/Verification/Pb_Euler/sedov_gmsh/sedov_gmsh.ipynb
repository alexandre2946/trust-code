# Étude comparative du problème de Sedov sur différents maillages

```python
from trustutils.jupyter import run
run.introduction("W. Aboussi, K. Pons and E. Saikali")
```

## Description du Probleme:
Ce cas test de Sedov permet de tester la résolution des équations d'Euler compressible avec un gaz parfait pour une explosion cylindrique. De l'énergie est déposée au centre menant à une expansion violente. Une solution analytique est disponible en cylindrique et en sphérique. Ici seul le cas cylindrique est vérifié. 
Une sensibilité au choix de maillage est présentée pour un solveur type-Godunov.



```python
run.TRUST_parameters()
mesh_conf = {"mesh_cart_0", "mesh_cart" , "mesh_struc_tetra", "mesh_hybrid","mesh_tetra"}
#mesh_conf = {"mesh_cart_0", "mesh_cart"}
```

```python
run.initBuildDirectory()
run.reset()
for msh in mesh_conf :      
    name = msh
    nb_post = 50
    substitutions_dict = {"mesh" : "../"+msh+".med", "nb" : nb_post}
    
    r = run.addCaseFromTemplate(f"jdd.data", targetDirectory=f"{name}", dic=substitutions_dict)        

run.printCases()
run.runCases()
perf = run.tablePerf() 
display(perf)
```

```python
# Recup valeurs dans fichier exterieur (resultats de "reference")
with open('src/cylindrical-sedov.out') as f1:
    for _ in range(19):
        next(f1)  # Ignore la ligne
        
    lines = f1.readlines()
    x_neg = [-float(line.split()[1]) for line in lines]
    x_pos = [float(line.split()[1]) for line in lines]
    x_1 = x_neg[::-1] + x_pos
    rho = [float(line.split()[2]) for line in lines]
    rho_1 = rho[::-1] + rho
```

```python
from trustutils import plot

sonde = plot.Graph("", nX= 2,nY= 2)
for msh in mesh_conf :            
    sonde.addPlot([0,0], "Numerical radial density x")
    sonde.addSegment(f"{msh}/jdd_RHO_X.son", time=1, label= f"{msh}")
    sonde.label('x [m]', 'rho [$kg.m^{-3}$]')
sonde.add(x_1, rho_1, label = 'reference solution', marker='--')      
for msh in mesh_conf :      
    sonde.addPlot([0,1], "Numerical radial density y")
    sonde.addSegment(f"{msh}/jdd_RHO_Y.son", time=1, label= f"{msh}")
    sonde.label('x [m]', 'rho [$kg.m^{-3}$]')
sonde.add(x_1, rho_1, label = 'reference solution', marker='--')   

for msh in mesh_conf :      
    sonde.addPlot([1,0], "Numerical radial density xy")
    sonde.addSegment(f"{msh}/jdd_RHO_XY.son", time=1, label= f"{msh}")
    sonde.label('x [m]', 'rho [$kg.m^{-3}$]')   
sonde.add(x_pos, rho, label = 'reference solution', marker='--')          

sonde.addPlot([1,1], "Numerical radial density xy")
sonde.addSegment(f"mesh_cart_0/jdd_RHO_X.son", time=1, label= f"cart 2500")
sonde.addSegment(f"mesh_cart/jdd_RHO_X.son", time=1, label= f"cart 20449")
sonde.label('x [m]', 'rho [$kg.m^{-3}$]')   
sonde.add(x_1, rho_1, label = 'reference solution', marker='--')          
```
