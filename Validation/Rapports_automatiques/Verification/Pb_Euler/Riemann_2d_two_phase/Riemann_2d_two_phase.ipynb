# 2D Two-Phase Riemann Problem

```python
from trustutils.jupyter import run
run.introduction("W. Aboussi, K. Pons and E. Saikali")
```

```python
run.TRUST_parameters()
run.reset()
```

```python
tests = {"galon", "fraysse"}
nx = 200 ; ny= 200

run.reset()
run.initBuildDirectory()
substitutions_dict = {"nx" : nx, "ny": ny}

for test in tests :          
    r = run.addCaseFromTemplate(f"jdd_{test}.data", targetDirectory=f"{test}", dic=substitutions_dict)        
    
run.printCases()
run.runCases()
run.tablePerf()
```

# Diagonal profiles

```python
from trustutils import plot
variables = ["RHO", "P", "V_L", "V_S"]

for test in tests:
    a = plot.Graph(title=f"Test {test}",nX=2, nY=2)
    nY = 0
    for i, p in enumerate(variables):
        iX = i % 2        # colonne (0 ou 1)
        iY = i // 2       # ligne (0 ou 1)
        
        a.addPlot((iX, iY), f"Profil de {p}")
        a.y_label = f"{p}"
        a.addSegment(f"{test}/jdd_{test}_{p}.son", label=f"{test}/{p}/solide",lw = 2, compo=0)
        a.addSegment(f"{test}/jdd_{test}_{p}.son", label=f"{test}/{p}/liquide",lw = 2, compo=1)
        
```
