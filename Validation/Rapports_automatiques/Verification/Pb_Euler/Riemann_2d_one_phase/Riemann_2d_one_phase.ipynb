# 2D One-Phase Riemann Problem

```python
from trustutils.jupyter import run
run.introduction("W. Aboussi, K. Pons and E. Saikali")
```

```python
run.TRUST_parameters()
run.reset()
```

```python

tests = {"t2", "t1"}
nx = 250 ; ny= 250

run.reset()
run.initBuildDirectory()
substitutions_dict = {"nx" : nx, "ny": ny}

for test in tests :      
    r = run.addCaseFromTemplate(f"jdd_{test}.data", targetDirectory=f"{test}", dic=substitutions_dict)        
    
run.printCases()
run.runCases()
run.tablePerf()
```

# Diagonal profiles

```python
from trustutils import plot
variables = ["RHO", "P", "V"]

for test in tests:
    a = plot.Graph(title=f"Test {test}",nX=1, nY=3)
    nY = 0
    for p in variables:
        a.addPlot(nY, f"Profil de {p}")
        a.addSegment(f"{test}/jdd_{test}_{p}.son", label=f"{test}/{p}",lw = 2)
        nY = nY+1 
```
